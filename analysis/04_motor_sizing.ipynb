# 04 Motor Sizing

Transparent first-pass sizing calculation with every input labelled as an assumption.

```python
import math
import pandas as pd
mass = 45.0 + 5.0
v = 0.8
r = 0.05
mu = 0.04
a = 0.4
g = 9.81
eta = 0.90
force = mass*a + mu*mass*g
torque = force*r/eta
power = force*v/eta
roller_omega = v/r
motor_rpm = 1750.0
ratio = (motor_rpm*2*math.pi/60)/roller_omega
print(pd.DataFrame([{'quantity':'force_N','value':force},{'quantity':'torque_Nm','value':torque},{'quantity':'power_W','value':power},{'quantity':'gear_ratio','value':ratio}]))
```

This is not a vendor selection. Real sizing would include belt pretension, startup load, roller and gearbox losses, duty cycle, service factor, thermal limits, braking and safety requirements.
